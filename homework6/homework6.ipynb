```python
arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])
def isprime (number):
    if number<=1:
        return False
    for i in range (2, number):
            if number%i ==0:
                return False
    return True
def rowprime (array):
    return array[np.any(np.vectorize(isprime)(array), axis=1)]

print (rowprime(arr))
```

```python
def blankboard():
    board = np.zeros((8,8))
    return board

def checkerboard():
    board = blankboard()
    board[::2,::2]=1
    board[1::2,1::2]=1
    return board
print(checkerboard())
def EVILcheckerboard():
    board = blankboard()
    board[1::2,::2]=1
    board[::2,1::2]=1
    return board
print(EVILcheckerboard())
```

```python
def expansion(word, number):
    res=""
    for i in word:
        res=res+i
        res=res+(" "*number)
    return res.rstrip()
expansionv = np.vectorize(expansion)
```
